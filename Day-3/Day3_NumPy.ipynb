```python
hours = np.array([2,4,6,3,7,5,1,8])
```

```python
average = np.mean(hours)
print("Average study hours:", average)
```

```python
max_hours = np.max(hours)
min_hours = np.min(hours)
print("Max study hours:", max_hours)
print("Min study hours:", min_hours)
```

```python
high_performers = hours[hours > 5]
print("Students who studied more than 5 hours:", high_performers)
print("Count:", len(high_performers))
```

```python
doubled_hours = hours * 2
print("Doubled study hours:", doubled_hours)
```

```python
x = np.linspace(0,5,11) #Creating 2 arrays with linear spacing.
x
y = x **2
y
```

```python
df = sns.load_dataset("tips")
df
```

```python
sns.displot(df['total_bill'],bins = 20)
```

```python
#Hue is used to divide them categorically on the basis of sex.
sns.pairplot(df,hue='sex')
```

```python
sns.countplot(x = df['sex'],hue = df['smoker'])
```

```python
#Bar Plots: In this y axis par quantitative will go and on x axis qualitative will go.
sns.barplot(x = df['sex'],y = df['total_bill'])
```

```python
tipscorr  = df[['total_bill','tip','size']]
sns.heatmap(tipscorr.corr(),annot= True)
```

```python
# y axis par jo predict karna h, and x axis par vo jiske through aapko predict karna h.
#Through this linear regression line will form.
sns.lmplot(x = 'total_bill',y = 'tip',data = df,hue = 'sex')
```
